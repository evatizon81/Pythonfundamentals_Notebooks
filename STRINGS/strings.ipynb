*Formatting → Formateo
Los strings permiten una operación llamada formatting, que sirve para insertar valores dentro de un texto.

1-Los %s son huecos donde Python coloca los valores.

```python
'%s, geranio, %s son las flores de mi jardin' % ('rosa','amapola')
```

2-Formatting con .format()
Otra forma es utilizar .format():
Los {} indican dónde queremos colocar los valores.

```python
'{}, geranio, {} son la flores de mi jardin'.format('rosa','amapola')
```

3-3. Formatting con f-string


```python
f'Las flores de mi jardin son {flores}'
```

4. .format() utilizando una posición

```python
flores = 'rosas y geranios'
'{0} son las flores'.format(flores)
```

*String Conversion Tools → Herramientas de conversión de strings

1. No podemos sumar un string y un número. 
2. string + número

En Python existen dos funciones que permiten convertir un carácter en su código numérico y viceversa: (código ASCII)

ord() → convierte un carácter en un número.



```python
ord('A')
```

chr() → convierte un número en su carácter correspondiente.

```python
chr(65)
```
