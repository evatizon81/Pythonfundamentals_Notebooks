STRINGS FUNDAMENTALS IN PYTHON
===============================


Es un conjunto de caracteres que se utiliza para representar texto.En Python, los strings se escriben entre comillas simples ' ' o dobles " ".
"Eva" y 'Hola mundo' son strings.

Crear una cadena de 4 caracteres y asignala a un nombre

```python
flores = "rosa"
print(flores)
```

longitud de una cadena

```python
len(flores)
```

Acceder a un caracter mediante el índice de la cadena

```python
flores[2]
```

El segundo elemento empezando por la izquierda.



```python
flores = "rosa"
flores[1]
```

El último elemento desde el final en S

```python
flores = "rosa"
flores[3]
```

El penultimo elemento

```python
flores = "rosa"
flores[2]
```

Backus-Naur

#flores[len(flores) -1] al añdir -1 a la longitud de la cadena, obtenemos el último índice

```python
flores[len(flores) -1]
```

Un segmento de flores

#flores[inicio:fin] incluye inicio, pero excluye fin.

```python
flores[1:3]
```

Concatenación y repetición

 "flores" + "arbol" = "floresarbol" lo que hace es unir las dos cadenas en una sola 

```python
"flores" + "arbol"
```
