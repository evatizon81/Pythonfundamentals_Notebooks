1. Operaciones básicas con listas
Las listas son secuencias, igual que las cadenas (str). Por eso, comparten algunas operaciones.

```python
len([1, 2, 3])  # 3
```

- len() es una función incorporada de Python.
- [1, 2, 3] es una lista con tres elementos.
- El resultado es 3 porque cuenta cuántos elementos tiene la lista.
Importante: len() cuenta los elementos, no sus valores ni sus índices.

Por ejemplo, len([10, 20, 30, 40]) devuelve 4.

```python
len(['one', [2, 3], 'four'])  # 3
```

len() cuenta los elementos del nivel de la lista que le pasamos, no los elementos que hay dentro de sus listas anidadas.
Tenemos estos 3 elementos:
1. 'one' → primer elemento.
2. [2, 3] → segundo elemento. Aunque contiene dos números, cuenta como un solo elemento de la lista principal.
3. 'four' → tercer elemento.

Las listas responden a los operadores + y * de forma muy similar a las cadenas de texto (strings).

# Concatenación

```python
[1, 2, 3] + [4, 5, 6]  # [1, 2, 3, 4, 5, 6]
```

* el operador + une las dos listas en una sola, colocando los elementos de la segunda al final de la primera.

# # Repeticion

```python
['Ni!'] * 4  # ['Ni!', 'Ni!', 'Ni!', 'Ni!']
```

el operador * repite la lista 4 veces.
- ['Ni!']: lista con un único elemento.
- * 4: repite ese elemento cuatro veces.

El operador + funciona igual para las listas y las cadenas de texto (strings). Es importante saber que espera el mismo tipo de secuencia en ambos lados; de lo contrario, se obtiene un error de tipo (TypeError).

```python
str([1, 2]) + "34"  # Igual que "[1, 2]" + "34"
```

str([1, 2]) convierte la lista [1, 2] en una cadena de texto: "[1, 2]".

El operador + une las dos cadenas de texto.

```python
[1, 2] + list("34")  # Igual que [1, 2] + ["3", "4"]
```

list("34") convierte la cadena "34" en una lista con dos elementos: ["3", "4"].

El operador + une las dos listas.

* List Iteration and Comprehension - Iteración y comprensión de listas.

En Python, la iteración permite recorrer los elementos de una lista, por ejemplo, mediante un bucle for.

La comprensión de listas permite crear una lista nueva a partir de otra secuencia, aplicando una operación o una condición de forma compacta.

. Comprobar si un elemento está en una lista.

```python
3 in [1, 2, 3]  # True
```

- in: comprueba si un elemento está dentro de una lista.
- 3: es el elemento que buscamos.
- [1, 2, 3]: es la lista donde lo buscamos.
- True: significa que el elemento sí está en la lista.

```python
for x in [1, 2, 3]:
    print(x, end=' ')
```

for x in [1, 2, 3]: recorre la lista y asigna cada elemento a la variable x.
print(x, end=' '): muestra el valor de x y añade un espacio al final, en lugar de pasar a una nueva línea.
* end=' ' hace que los números aparezcan en la misma línea, separados por espacios. Sin end=' ', cada número aparecería en una línea diferente.

* Las comprensiones de listas (list comprehensions) son una forma de crear una lista nueva aplicando una expresión a cada elemento de una secuencia.

```python
res = [c * 4 for c in 'SPAM']
res
# ['SSSS', 'PPPP', 'AAAA', 'MMMM']
```

for c in 'SPAM': recorre la cadena de texto carácter por carácter (S, P, A, M).
c * 4: repite cada carácter cuatro veces.
[ ... ]: crea una lista con los resultados.
res =: guarda la lista resultante en la variable res.

* Este código es equivalente a este otro:

```python
# CRUD
res = []                     # Create (Crear)

for c in 'SPAM':             # Read (Leer)
    res.append(c * 4)         # Update (Actualizar)

del res[0]                   # Delete (Eliminar)

res
# ['PPPP', 'AAAA', 'MMMM']
```

CRUD en listas de Python
CRUD representa cuatro operaciones básicas que podemos realizar con los datos: Create, Read, Update y Delete.
* - Create (Crear): res = [] crea una lista vacía.
- Read (Leer): for c in 'SPAM' recorre la cadena carácter por carácter.
- Update (Actualizar): res.append(c * 4) añade a la lista cada carácter repetido cuatro veces.
- Delete (Eliminar): del res[0] elimina el elemento situado en la posición 0, es decir, 'SSSS'.

* La función integrada map() realiza un trabajo similar, pero aplica una función a los elementos de una secuencia y recopila todos los resultados en una lista nueva.

```python
list(map(abs, [-1, -2, 0, 1, 2]))
# [1, 2, 0, 1, 2]
```

abs: es una función que devuelve el valor absoluto de un número, es decir, su valor sin signo negativo.
map(abs, [...]): aplica la función abs() a cada elemento de la lista.
list(...): convierte el resultado de map() en una lista.

#* Indexing, Slicing, and Matrixes - Indexación, segmentación y matrices

En Python:
- Indexación: permite acceder a un elemento concreto mediante su posición, por ejemplo, lista[0].
- Segmentación: permite obtener una parte de una lista, por ejemplo, lista[1:3].
- Matrices: permiten organizar datos en filas y columnas, normalmente mediante listas anidadas.

```python
L = ['spam', 'Spam', 'SPAM!']

L[2]  # 'SPAM!'
```

L es una lista con tres elementos.
L[2] accede al elemento que está en la posición 2.
En Python, los índices empiezan por 0, no por 1.

```python
L[-2]               # 'Spam'
```

*Los índices negativos permiten acceder a los elementos de una lista empezando por el final.

```python
L[1:]               # ['Spam', 'SPAM!']
```

* L[1:] es una segmentación (slicing) que obtiene una parte de la lista.
- 1: índice donde comienza la selección.
- :: indica que continúa hasta el final de la lista.
Recuerda que los índices empiezan en 0, por lo que se seleccionan los elementos desde el segundo hasta el último.


```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
matrix[1]
```

* matrix es una lista que contiene otras tres listas (filas).
- matrix[0] → [1, 2, 3] (primera fila).
- matrix[1] → [4, 5, 6] (segunda fila).
- matrix[2] → [7, 8, 9] (tercera fila).
Importante: los índices empiezan en 0. Por eso, matrix[1] devuelve la segunda fila completa.

```python
matrix[2][0]
```

1. matrix[2] → selecciona la tercera fila: [7, 8, 9].
2. [0] → selecciona el primer elemento de esa fila: 7.

```python
matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]
matrix[1][1]
```


1. matrix[1] → selecciona la segunda fila: [4, 5, 6].
2. [1] → selecciona el segundo elemento de esa fila: 5.
Resultado: 5

* el primer índice selecciona la fila y el segundo selecciona el elemento dentro de esa fila.

#*Changing Lists in Place - Modificar listas directamente

Como las listas son mutables (mutable), permiten realizar operaciones que modifican la propia lista directamente.

Esto significa que las operaciones de esta sección modifican el objeto lista, sustituyendo su contenido anterior sin necesidad de crear una copia nueva.

* una lista es mutable porque podemos cambiar sus elementos sin crear otra lista.

Index and slice assignments - Asignaciones mediante índices y segmentos

```python
L = ['spam', 'Spam', 'SPAM!']
L[1] = 'eggs'     # Asignación por índice
L
# ['spam', 'eggs', 'SPAM!']
```

- L = ['spam', 'Spam', 'SPAM!']: creamos una lista con tres elementos.
- L[1] = 'eggs': sustituimos el elemento que está en el índice 1 por 'eggs'.
- L: mostramos la lista modificada.

* Importante: las listas son mutables, es decir, podemos modificar sus elementos sin crear una lista nueva. Esta operación se llama asignación por índice (index assignment).

```python
L[0:2] = ['eat', 'more']  # Asignación por rebanadas: borrar + insertar
L                         # ['eat', 'more', 'SPAM!']
```

L[0:2]: selecciona los elementos desde el índice 0 hasta el 2, sin incluirlo. Es decir, los elementos de los índices 0 y 1.
= ['eat', 'more']: sustituye esos elementos por 'eat' y 'more'.
L: muestra la lista modificada.
* el índice final de una rebanada no se incluye. Por eso, L[0:2] selecciona los índices 0 y 1, pero no el 2.

```python
L = [1, 2, 3]       # Creamos la lista
L[1:2] = [4, 5]     # Sustitución e inserción
L
# [1, 4, 5, 3]
```

L = [1, 2, 3]: creamos una lista con tres elementos.
L[1:2]: seleccionamos desde el índice 1 hasta el 2, sin incluirlo. Por tanto, seleccionamos únicamente el 2.
L[1:2] = [4, 5]: sustituimos el elemento 2 por dos elementos: 4 y 5.
L: mostramos la lista modificada.
* una asignación por rebanadas puede sustituir un elemento por varios. Por eso, la lista pasa de tener 3 elementos a tener 4.

```python
L[1:1] = [6, 7]  # Inserción: no sustituye ningún elemento
L
# [1, 6, 7, 4, 5, 3]
```

- L[1:1]: selecciona una rebanada vacía, porque empieza y termina en el índice 1.
- = [6, 7]: inserta los elementos 6 y 7 en esa posición.
- L: muestra la lista modificada.
* como L[1:1] no selecciona ningún elemento, no se borra ni se sustituye nada. Simplemente se insertan los nuevos elementos antes del que estaba en el índice 1.

```python
L[1:2] = []  # Eliminación: no inserta nada
L
# [1, 7, 4, 5, 3]
```

- L[1:2]: selecciona el elemento que está en el índice 1.
- = []: sustituye ese elemento por una lista vacía. Como no hay elementos que insertar, el elemento seleccionado se elimina.
- L: muestra la lista modificada.
* na asignación por rebanadas también permite eliminar elementos. Si asignamos una lista vacía ([]), borramos los elementos seleccionados sin insertar ninguno.

```python
L = [1]
L[:0] = [2, 3, 4]  # Inserta todos al principio
L
# [2, 3, 4, 1]
```

L = [1]: creamos una lista con el elemento 1.
L[:0]: selecciona una rebanada vacía al principio de la lista.
= [2, 3, 4]: inserta los elementos 2, 3 y 4 al principio, sin eliminar el 1.
L: muestra la lista modificada.
* L[:0] equivale a L[0:0]. Como la rebanada está vacía, no sustituye ningún elemento: simplemente inserta los nuevos al principio.

Recuerda que L[:0] = [...] permite añadir varios elementos al principio de una lista sin borrar los que ya existen.

```python
L[len(L):] = [5, 6, 7]  # Insert all at len(L):, an empty slice at end
L
```

len(L) obtiene la longitud de la lista.
L[len(L):] selecciona una porción vacía al final.
= [5, 6, 7] añade esos elementos al final de la lista.
* modifica la lista original, porque las listas son mutables.

#* List method calls - Llamadas a métodos de listas
* Al igual que las cadenas de texto (strings), los objetos de tipo lista de Python también admiten llamadas a métodos específicos, muchos de los cuales modifican la lista original directamente.
* Los métodos son funciones (en realidad, atributos de objetos que hacen referencia a funciones) que están asociadas a determinados objetos y actúan sobre ellos.

* append(): añade un elemento al final de una lista y modifica la lista original.

```python
L = ['eat', 'more', 'SPAM!']
L.append('please')  # Añade un elemento al final
L
```

* append() añade un elemento al final de la lista y modifica la lista original.

* Ten en cuenta que el método append() solo añade un elemento nuevo al final de la lista.
Si añades una lista a otra mediante append(), la lista añadida se convierte en un único elemento al final de la primera lista

```python
a = [1, 2, 3, 4, 5, 6, 7, 7]
b = [8, 9]
a.append(b)                     # a: [1, 2, 3, 4, 5, 6, 7, 7, [8, 9]]
a[8]
[8, 9]
```

* append(b) añade la lista b completa como un elemento de a.
* a[8] accede al elemento de índice 8, que es la lista [8, 9].
* Los índices empiezan en 0, por eso el elemento añadido está en la posición 8.

SORT 

```python
L.sort()  # Ordena los elementos de la lista ('S' < 'e')
L
```

* sort() ordena los elementos de la lista modificando la lista original.
* Python distingue entre mayúsculas y minúsculas; por eso, en este ejemplo, 'SPAM!' aparece antes que 'eat'.

 * En sort(), el argumento reverse permite ordenar en orden descendente en lugar de ascendente, y el argumento key recibe una función de un argumento que devuelve el valor que se utilizará para ordenar.
*reverse=True: ordena de mayor a menor.
*key: indica qué criterio se utiliza para ordenar.

```python
L = ['abc', 'ABD', 'aBe']  # Ordenar con mayúsculas y minúsculas mezcladas
L.sort()
L
```

* sort() ordena alfabéticamente, distinguiendo entre mayúsculas y minúsculas. Por eso, las palabras que empiezan por mayúscula aparecen antes que las que empiezan por minúscula.

```python
L = ['abc', 'ABD', 'aBe']  # L se modificó directamente con sort() anteriormente; ¡es un objeto mutable!
L.sort(key=str.lower)
L
```

* key=str.lower ordena las palabras sin distinguir entre mayúsculas y minúsculas.
* sort() modifica la lista original (in place).

```python
L = ['abc', 'ABD', 'aBe']
L.sort(key=str.lower, reverse=True)  # Cambia el orden de clasificación
L
```

* key=str.lower: ordena sin distinguir mayúsculas y minúsculas.
* reverse=True: invierte el orden, de descendente a ascendente.

* sort() modifica la lista original.

POP
El método pop() de las listas se utiliza a menudo junto con append() para implementar una estructura de pila LIFO (último en entrar, primero en salir). El final de la lista actúa como la parte superior de la pila.

```python
L = []
L.append(1)  # Añade 1 a la pila
L.append(2)  # Añade 2: [1, 2]
L.pop()      # Extrae el último elemento de la pila
L            # [1]
```

* append() añade elementos al final de la lista.
* pop() elimina y devuelve el último elemento, en este caso, el 2.
* L queda como [1].
* El último en entrar es el primero en salir.

Iterating over a list - Recorrer una lista
Python permite utilizar un bucle for directamente sobre una lista

```python
my_list = ['foo', 'bar', 'baz']

for item in my_list:
    print(item)
```

* my_list: mi lista.
* for item in my_list: recorrer la lista elemento por elemento.
* print(item): mostrar cada elemento por pantalla.

* Puedes obtener también la posición de cada elemento al mismo tiempo.
* Puedes recorrer una lista obteniendo el índice y el valor de cada elemento usando enumerate().

```python
for (index, item) in enumerate(my_list):
    print('El elemento en la posición {} es: {}'.format(index, item))
```

* enumerate(my_list) obtiene el índice y el elemento de cada posición.
* index guarda el índice (0, 1, 2...).
* item guarda el elemento (foo, bar, baz).
* .format(index, item) coloca esos valores dentro de las llaves {} del texto.

Any and All - Cualquiera y Todos
Usa all() para comprobar si todos los valores de un iterable se evalúan como True.
* all() devuelve True si todos los elementos son verdaderos.
Si al menos uno es falso, devuelve False.

```python
nums = [1, [1], 'a', 1]
all(nums)  # True (Verdadero)
```

* all() devuelve True porque todos los elementos se consideran verdaderos en Python:
1 → verdadero.
[1] → verdadero porque la lista no está vacía.
'a' → verdadero porque la cadena no está vacía.
1 → verdadero.
* all() no exige que los elementos sean literalmente True, solo que se evalúen como verdaderos.

* any() comprueba si uno o más valores de un iterable se evalúan como True.
* any() devuelve True si al menos un elemento es verdadero.
* Devuelve False si todos los elementos son falsos.

```python
nums = [1, 1, 0, 1]
any(nums)  # True (Verdadero)
```

* any() devuelve True porque al menos un elemento es verdadero.

```python
vals = [None, 0, [], False, '']
any(vals)  # False (Falso)
```

* any() devuelve False porque todos los elementos se consideran falsos en Python:
* None → ausencia de valor.
* 0 → cero.
* [] → lista v

*Estas funciones integradas funcionan con cualquier iterable, incluidos los generadores.
* all() y any() pueden trabajar con listas, tuplas y otros iterables, como los generadores.
* Un iterable es un objeto que permite recorrer sus elementos uno a uno.

```python
vals = [1, 2, 3, 4]
any(val > 12 for val in vals)  # False
```

* for val in vals recorre los números de la lista.
* val > 12 comprueba si cada número es mayor que 12.
* any() devuelve True si al menos uno cumple la condición.
* Como ningún número es mayor que 12, devuelve False.
