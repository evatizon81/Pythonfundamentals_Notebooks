OBJETOS INMUTABLES

En Python, los objetos de tipo integer (int) son inmutables, lo que significa que su valor no puede modificarse una vez creado el objeto. Sin embargo, una variable sí puede pasar a referenciar otro objeto.

```python
a = 6 
b = a
print("b es ", b )                  # b es 6

a = 6
b = a
a = 'spam'                          # b es 6
print("b es ", b)
```

- Los objetos Integer no son mutables.

```python
a = 6 
b = a
a= a + 2                    # b es 6
print ("b es ", b)
```

OBJETOS MUTABLES




. Listas



- Las listas son mutables, mientras que los enteros son inmutables.

```python
L = [2, 3, 4]               # Objeto mutable.
M = L                       # Referencia al mismo objeto.
L[0] = 24                   # Modificación del objeto.

print ("L = ", L)
print ("M = ", M)
```

- L y M apuntan a la misma lista.

- En el caso de L[0] : 
L es la lista 
[0] indica la posición del primer elemento(Py empieza a contar desde 0) 
= 24 sustituye el elemento que estaba en esa posición por '24'

* Por lo tanto la lista cambia:
[2,3,4] pasa a [24,3,4]
como M apunta a la misma lista vemos [24,3,4]


.Copia objetos en lugar de referencia:

```python
L = [2, 3, 4]
M = L[:]		    # Crea una copia de L
L[0] = 24

print("L = ", L)    # [24, 3, 4]

print("M = ", M)    # [2, 3, 4] M no cambia
L =  [24, 3, 4]
M =  [2, 3, 4]
```

- L[:]  significa seleccionar todos los elementos de la lista, desde el principio hasta el final.

Al utilizarla en una asignación: M = L[:] crea una lista nueva y asigna su referencia a M.

- Que ocurre cuando ejecutamos L[0] = 24?
Solo se modifica el primer elemento de la lista a la que apunta L.

Por tanto:
L contiene [24, 3, 4].
M contiene [2, 3, 4].

La modificación de L no afecta a M.



Shared References and Equality -  Referencias compartidas e igualdad.

Operador == igualdad de valores.
is comprueba si dos referencias apuntan al mismo objeto.

```python
L = [1, 2, 3]
M = L

print(L == M)  # True
print(L is M)  # True
```

L = [1, 2, 3] crea una lista.
M = L hace que M apunte a esa misma lista.
L == M devuelve True porque sus valores son iguales.
L is M devuelve True porque ambas variables hacen referencia al mismo objeto.

```python
L = [1, 2, 3]
M = [1, 2, 3]

print("L == M", L == M)
print("L is M", L is M)
```


L == M es True porque las dos listas contienen los mismos elementos en el mismo orden.
L is M es False porque cada asignación literal crea una lista diferente.

. CACHE
Los enteros pequeños y las cadenas de texto se almacenan en caché y se reutilizan, el operador is nos indica que hacen referencia a un mismo objeto.

```python

X = 42
Y = 42  # Deberían ser dos objetos diferentes

print(X == Y)
print(X is Y)
```

Por qué las dos expresiones dan True?

X == Y → True: ambas variables tienen el mismo valor, 42.
X is Y → normalmente True en este ejemplo: Python puede reutilizar el mismo objeto entero para las dos variables.

Aunque hayas escrito 42 dos veces, eso no garantiza que Python cree dos objetos diferentes.

- Averiguar el número de referencias a un objeto:

```python
import sys
sys.getrefcount(1)		# numero de referencias a este objeto compartido en memoria.
```
