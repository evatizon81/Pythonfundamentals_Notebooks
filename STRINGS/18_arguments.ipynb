* Arguments and Shared References - Argumentos y referencias compartidas
Paso de objetos inmutables

```python
def f(a):          # a se asigna al objeto recibido como argumento
    a = 99         # Solo cambia la variable local a

b = 88
f(b)               # a y b hacen referencia inicialmente al mismo objeto 88
print("b = ", b)   # b no cambia
```


* los números son inmutables en Python. Al hacer a = 99, cambiamos a qué valor hace referencia a, pero b sigue valiendo 88.
- b = 88: creamos la variable b.
- f(b): pasamos el valor de b a la función, donde se llama a.
- a = 99: cambiamos el valor de la variable local a, no el de b.
- print(b): sigue mostrando 88.


Paso de objetos mutables:

```python
def changer(a, b):
    a = 2              # Solo cambia la variable local a
    b[0] = 'spam'      # Modifica la lista original

x = 1
l = [1, 2]

changer(x, l)

print("l = ", l)       # ['spam', 2]
assert x == 1, "x es 1"
```

* los objetos mutables, como las listas, pueden modificarse dentro de una función y esos cambios se ven fuera de ella.
- x = 1: es un número inmutable. Cambiar a no modifica x.
- l = [1, 2]: es una lista mutable.
- a = 2: solo cambia la variable local a.
- b[0] = 'spam': modifica el primer elemento de la lista original.
Resultado: x sigue valiendo 1, pero l pasa a ser ['spam', 2].

Cómo evitar que una rutina modifique los elementos mutables:
- Pasar una copia

```python
l = [1, 2]
changer(x, l[:])  # Pasamos una copia para que l no cambie
print("l = ", l)
```

Explicación breve:
- l[:] crea una copia de la lista.
- changer() modifica esa copia, no la lista original.
- Por eso, l sigue siendo [1, 2].
Clave: l pasa la lista original; l[:] pasa una copia.

- Utilizar invariantes (en las estructuras de datos de entrada)

```python
l = [1, 2]

def changer(a, b):
    b = b[:]           # Copia la lista recibida para no modificar la original
    a = 2
    b[0] = 'spam'      # Modifica únicamente la copia

changer(x, l)

assert l == [1, 2], "l es una invariante"
print("l = ", l)
```

* Al copiar la lista dentro de la función, evitamos modificar la lista original.

- b = b[:]: crea una copia de la lista.
- b[0] = 'spam': modifica solo la copia.
- l permanece igual: [1, 2].


*- Convertir el objeto mutable a un objeto inmutable:

```python
l = [1, 2]

def changer(a, b):
    a = 2
    b[0] = 'spam'  # Error: no se puede modificar una tupla

changer(x, tuple(l))  # Pasamos una tupla para impedir modificaciones
print("l = ", l)
assert l == [1, 2], "l es una invariante"
```

* Las listas son mutables; las tuplas son inmutables.
- tuple(l): convierte la lista l en una tupla.
- Las tuplas son inmutables, por lo que no se puede cambiar b[0].
- Python genera un TypeError al intentar modificar la tupla.
- La lista original l sigue siendo [1, 2].
