
ÁMBITO GLOBAL (GLOBAL SCOPE)
- Los nombres globales son variables asignadas en el nivel superior del archivo del módulo que las contiene.
- Los nombres globales solo deben declararse si se les asigna un valor dentro de una función.
- Los nombres globales pueden utilizarse dentro de una función sin necesidad de declararlos.


```python
x = 88          # X global

def func():
    x = 99      # X local: oculta la global, pero queremos esto aquí

func()
print(x)        # Imprime 88: no ha cambiado
```

-Cuando asignas un valor a x dentro de una función, Python considera que x es local, salvo que declares lo contrario con global.

x = 88: crea una variable global, accesible desde distintas partes del módulo.
def func():: define una función.
x = 99: crea una variable local dentro de la función. No modifica la variable global.
func(): ejecuta la función.
print(x): imprime 88, porque la variable global sigue teniendo su valor original.

```python
x = 88          # X global

def func():
    global x
    x = 99      # X global: fuera de la función

func()
print("x =", x)  # Imprime 99
```

Traducción y explicación
- x = 88: X global.
- global x: indica que vamos a utilizar la variable global x.
- x = 99: modifica la variable global, aunque estemos dentro de la función.
- func(): ejecuta la función.
- print("x =", x): imprime x = 99.
Diferencia importante:
- Sin global x, x = 99 crea una variable local.
- Con global x, x = 99 modifica la variable global.
Por eso, al final, el valor de x es 99.












LEGB rule


Es el orden que sigue Python para buscar los nombres de las variables:
- L — Local: dentro de la función.
- E — Enclosing: dentro de una función que contiene a otra.
- G — Global: en el módulo.
- B — Built-in: nombres incorporados de Python, como print().


```python
y, z = 1, 2          # Variables globales del módulo

def all_global():
    global x         # Declarar las variables globales a las que se asigna un valor
    x = y + z        # No hace falta declarar y, z: regla LEGB

all_global()

print('x = %d, y = %d, z = %d' % (x, y, z))
```

y, z = 1, 2: crea dos variables globales, con valores 1 y 2.
global x: indica que x será una variable global porque vamos a asignarle un valor dentro de la función.
x = y + z: suma y + z y guarda el resultado (3) en x.
No hace falta declarar y ni z con global, porque solo estamos leyendo sus valores, no modificándolos.
all_global(): ejecuta la función.

print(...): muestra las tres variables.

NESTED SCOPES - AMBITOS ANIDADOS

- Hay ámbitos dentro de otros ámbitos, por ejemplo, una función definida dentro de otra función.
- Nested = anidado.
- Scopes = ámbitos o alcances de las variables.
En Python, una función interna puede acceder a las variables de la función externa que la contiene.

```python
x = 99                       # Variable del ámbito global

def f1():
    x = 88                   # Variable local de f1

    def f2():                # Función anidada dentro de f1
        print("x local = ", x)  # Referencia a la variable del ámbito externo

    f2()                     # f2 solo existe dentro de f1

f1()                          # Imprime 88
print("x global = ", x)       # Imprime 99
```

Explicación paso a paso
1. x = 99: crea una variable global.
2. f1() crea una variable local x = 88.
3. Dentro de f1() se define f2().
4. f2() utiliza x, pero no tiene una variable x propia. Python la busca en la función que la contiene (f1) y encuentra el valor 88.
5. Cuando termina f1(), se imprime la variable global, que sigue valiendo 99.

* f2() solo se puede llamar desde dentro de f1(), porque está definida dentro de ella.

Factory Functions: Closures - Funciones fábrica: cierres.

*Factory function: función que crea y devuelve otra función.

*Closure: función interna que recuerda las variables de la función externa, incluso cuando esta ya ha terminado.

```python
def f1():
    x = 88                 # Ámbito de la función externa

    def f2():
        print("x =", x)    # Recuerda x del ámbito de la función externa

    return f2              # Devuelve f2, pero no la ejecuta
                           # f1() devuelve el objeto función f2

action = f1()              # Crea y devuelve la función; action pasa a ser f2
action()                   # La ejecuta: imprime 88, equivale a f2()
```

- x = 88: variable de la función externa.
- return f2: devuelve la función, pero no la ejecuta.
- action = f1(): guarda la función devuelta en action.
- action(): ejecuta esa función e imprime 88.

* aunque f1() haya terminado, f2() recuerda el valor de x.

Eso es una closure: una función que recuerda las variables de su ámbito externo, incluso después de que la función externa haya terminado.

```python
def maker(n):
    def action(x):       # Crear y devolver action
        return x ** n    # action conserva n del ámbito externo
    return action

f = maker(2)             # Pasar 2 al argumento n → devuelve x ** 2
f(3)                     # Pasar 3 a x; n recuerda el valor 2 → devuelve 3 ** 2
f(4)                     # Devuelve 4 ** 2
```

f = maker(2)
- maker(2) guarda el valor 2 en n.
- Devuelve la función action.
- Por tanto, f es ahora esa función y recuerda que n = 2.
f(3)
Es como ejecutar action(3):     #return 3 ** 2 - resultado : 9
f(4)
Es como ejecutar action(4):     #return 4 ** 2 - resultado : 16.


* ** significa elevar a una potencia. Por ejemplo, 4 ** 2 es \(4 \times 4 = 16\).
aunque maker() ya haya terminado, la función f recuerda que n = 2. Por eso eleva al cuadrado los números que le pasamos.
