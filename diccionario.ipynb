# DICCIONARIOS

*Los items se almacenan y recuperar por una clave, en vez de por su posición en la estructura.

*Es una tipo de agregación en el cual el nombre de los items tiene más significado que su posición.

*Los diccionarios pueden reemplazar muchos de los algoritmos de búsqueda y estructuras de datos que deberías implementar manualmente en lenguajes de bajo nivel.

Características:

Se accede a los elementos por su clave o key, no por la posición que ocupan.
Son colecciones no ordenadas de objetos de cualquier tipo, y el tipo de los elementos que lo forman puede ser distinto.
Poseen longitud variable y pueden anidarse.
Son mapas mutables.
Son tablas de referencias a objetos o hash tables.

1. ¿Qué es un diccionario?

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
```

- Los valores son los datos asociados a cada clave.
- Se utilizan {} para crear el diccionario y : para separar clave y valor.

# Cambiando diccionarios in place

D['ham'] = ['grill', 'bake', 'fry']         # Change entry (value=list)
D

* Elimina la entrada cuya clave es 'eggs' del diccionario D.

# Métodos de un objeto diccionario
values

El método values devuelve todos los valores del diccionario.

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
list(D.values())                
# [3, 2, 1]
```

* Asigna un nuevo diccionario a D y values() devuelve una lista con todos sus valores: [3, 2, 1].

* Items
items devuelve todas las tuplas (key,value).

```python
list(D.items())
# [('eggs', 3), ('spam', 2), ('ham', 1)]
```

* items() devuelve una lista con todos los pares clave-valor del diccionario.

* get

Fetching una clave que no existe es normalmente un error, pero el método get devuelve None como valor por defecto o un valor pasado como por defecto si la clave no existe.

```python
D['toast']      # KeyError: 'toast'   
```

* Da un error KeyError porque la clave 'toast' no existe en el diccionario.

```python
D.get('spam')
# 2
```

* get() devuelve el valor asociado a la clave 'spam': 2.

```python
print(D.get('toast'))       # Una clave que no existe
# None
```

* get() devuelve None si la clave 'toast' no existe, sin producir un error.

```python
D.get('toast', 88)          # proporcionando un valor por defecto
# 88
```

* get() devuelve 88 como valor por defecto si la clave 'toast' no existe.

update

El metodo update une las claves y valores de un diccionaroi en otro, sobreescribiendo valores de la misma clave si hay conflicto.

```python
# D = {'eggs': 3, 'spam': 2, 'ham': 1}
D2 = {'toast':4, 'muffin':5, 'ham': 10}
D.update(D2)
D
# {'eggs': 3, 'muffin': 5, 'toast': 4, 'spam': 2, 'ham': 10}
```

* update() añade las nuevas claves de D2 a D y sustituye el valor de 'ham' por 10, porque esa clave ya existía.

* pop

pop elimina una clave de un diccionario y devuelve el valor asociado a la clave.

```python
D.pop('muffin')
# 5
D.pop('toast')
# 4
D
# {'eggs': 3, 'spam': 2, 'ham': 10}

```

* pop() devuelve el valor de la clave indicada y elimina esa entrada del diccionario.
* pop() obtiene y elimina las claves 'muffin' y 'toast', devolviendo sus valores (5 y 4).

# Ejemplos

* Movie Database

```python
table = {'1975': 'Holy Grail', '1979': 'Life of Brian', '1983': 'The Meaning of Life'}
year = '1983'
movie = table[year]
movie
# 'The Meaning of Life'
```

* Crea un diccionario table y utiliza la variable year para buscar la película correspondiente a 1983, guardándola en movie.

```python
for year in table:                      # Los mismo que: for year in table.keys()
    print(year + '\t' + table[year])
```

* Recorre todas las claves del diccionario e imprime cada año junto con su película correspondiente.

* Para cualquier diccionario D, for key in D funciona de la misma manera que for key in D.keys()

```python
for key in table:
    print(key + '\t' + table[key])
```

* Recorre las claves del diccionario y muestra cada clave (key) junto con su valor (table[key]).

```python
for key in table.keys():
    print(key + '\t' + table[key])
```

* Recorre explícitamente las claves del diccionario con keys() e imprime cada clave junto con su valor.

* Mapeando valores a claves:

```python
table = {'Holy Grail': '1975', 'Life of Brian': '1979', 'The Meaning of Life': '1983'}
table['Holy Grail']
```

* Accede al valor asociado a la clave 'Holy Grail', que es '1975'.

```python
list(table.items())         # lista de tuplas (clave, valor)
```

* Devuelve una lista de tuplas, cada una con una pareja de clave y valor del diccionario.

# Comprehension syntax

```python
[title for (title, year) in table.items() if year == '1975']
```

*Crea una lista con los títulos de las películas cuyo año sea '1975', recorriendo los pares clave-valor del diccionario mediante items().

```python
V = '1975'
[key for (key, value) in table.items() if value == V]
```

* Busca en el diccionario las claves cuyo valor coincida con V ('1975') y devuelve una lista con esas claves.

```python
[key for key in table.keys() if table[key] == V]
```

* Crea una lista con las claves cuyo valor en el diccionario coincida con V ('1975'). Es otra forma de hacer la búsqueda anterior.

# Ordenando claves

```python
Ks = D.keys()
Ks = list(Ks)
Ks.sort()
for k in Ks:
    print(k, D[k])
```

* Obtiene las claves del diccionario, las convierte en una lista y las ordena alfabéticamente con sort(). Después, recorre las claves e imprime cada una junto con su valor correspondiente.

```python
D = {'b': 2, 'c': 3, 'a': 1}
Ks = D.keys()
for k in sorted(Ks): 
    print(k, D[k])
```

* Sección 16.6: Iterar sobre diccionarios
Considerando el siguiente diccionario:

d = {"a": 1, "b": 2, "c": 3}

1. Recorrer las claves:

```python
for key in d:
    print(key)
```

* Recorre e imprime todas las claves del diccionario. Es equivalente a for key in d.keys().

2. Recorrer los valores:

```python
for value in d.values():
    print(value)
```

* Recorre e imprime todos los valores del diccionario.

3. Recorrer claves y valores:

```python
for key, value in d.items():
    print(key, "::", value)
```

* Recorre los pares clave-valor y permite acceder a ambos elementos en cada iteración.

Diferencia entre Python 2 y Python 3:
- Python 2: keys(), values() e items() devuelven listas. Los métodos iterkeys(), itervalues() e iteritems() generan los elementos uno a uno.
- Python 3: keys(), values() e items() devuelven vistas dinámicas, no listas. Los métodos iterkeys(), itervalues() e iteritems() ya no existen.

* En Python 3.7 y posteriores, los diccionarios mantienen el orden de inserción de sus claves. Por tanto, el último comentario del texto (que indica que no siguen ningún orden) está desactualizado.
